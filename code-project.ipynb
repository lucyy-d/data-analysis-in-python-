```python
#install enviroment
import os
import numpy as np
import pandas as pd

# download file
FILE_NAME = "global_retail_prices_dataset.csv"
if not os.path.exists(FILE_NAME):
    FILE_NAME = "projekt/global_retail_prices_dataset.csv"

df = pd.read_csv(FILE_NAME)
print("downloaded\n")

# STEP 1: SHAPE
print("STEP 1: SHAPE")
print(df.shape)

# STEP 2: TYPES
print("STEP 2: DATA TYPES")
print(df.dtypes)

if "date_recorded" in df.columns:
    df["date_recorded"] = pd.to_datetime(df["date_recorded"], errors="coerce")
    print("\nAfter conversion 'date_recorded' -> datetime64:")
    print(df["date_recorded"].dtype)
print("\n")

# STEP 3: HOW MISSING VALUES ARE CODED
print("STEP 3: MISSING VALUES")

# 1.  Standard Lacks (NaN / null)
print("1. Standard lacks (NaN/null):")
print(df.isnull().sum())

# 2. Other lacks (puste stringi, 'N/A', 'null', '?', itp.)
print("\n2. Other lacks:")
text_columns = df.select_dtypes(include=["object", "string"]).columns

for col in text_columns:
    hidden_missing = (
        df[col]
        .isin(
            [
                "",
                " ",
                "N/A",
                "NA",
                "null",
                "NULL",
                "None",
                "unknown",
                "Unknown",
                "?",
            ]
        )
        .sum()
    )
    print(
        f"Column '{col}': {hidden_missing} other types of lacks"
    )
print("\n")

# STEP 4: GROUPS MIXED IN WITH UNITS
print("STEP 4: UNITS & CURRENCIES")

# Checking the currency 
if "currency" in df.columns:
    print(df["currency"].value_counts(dropna=False))

# checking cities 
if "city" in df.columns:
    print(df["city"].value_counts(dropna=False))
print("\n")

# STEP 5: CATEGORICAL VALUES & INCONSISTENCIES
print("STEP 5: CATEGORICAL VALUES")

# Verification of unique categories and consistency of entries
if "category" in df.columns:
    print(f"Number of unique categories: {df['category'].nunique()}")
    print("\nValues in column 'category' (including lacks):")
    print(df["category"].value_counts(dropna=False))

if "store_name" in df.columns:
    print(
        f"\nNumber of unique stores (store_name): {df['store_name'].nunique()}"
    )
print("\n")

# STEP 6: DUPLICATES & OUTLIERS CHECK
print("STEP 6: DUPLICATES & OUTLIERS")

# 1. Checking for complete duplicate rows
duplicate_rows = df.duplicated().sum()
print(f"Number of duplicated rows : {duplicate_rows}")

# 2. Rapid verification of extreme values ​​/ anomalies in prices
if "price" in df.columns:
    print("\nDescriptive statistics for prices (checking min, max, and quartiles):")
    print(df["price"].describe())
```
