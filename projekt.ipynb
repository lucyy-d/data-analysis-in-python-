## ***The dataset***

- Source :Open Food Facts – *Open Prices*, a non-profit, crowdsourced database of real shelf prices and receipts. Licence: Open Database License (ODbL).
- Link :  https://huggingface.co/datasets/openfoodfacts/open-prices (project website: https://prices.openfoodfacts.org)
- What one row is :  one price observation, meaning the price of one product (identified by its EAN/GTIN barcode in product_code) seen in one store on one day, as submitted by one contributor.  Unit price is the price of one item or package as labelled on the shelf, in local currency.
- the dataset covers Years from 2009-04-09 to 2026-10-29
- and covers 16,216,800 units, where there are 3,734,416 missing data points

## ***Why it matters***

- Households: if the same products are consistently cheaper in one chain, the choice of store is a real way to save money, especially for low-income families who spend a larger share of their budget on food.
- Competition authorities: under competition, identical goods should sell at similar prices (the law of one price). Large, persistent gaps point to market power, search costs or price discrimination by store format. This can be usefull information to  competition investigations and other competitors.
- Retailers and suppliers: a chain can see where it is priced above rivals on branded products and adjust its pricing strategy in order not to loose their clients, or to attract new potential buyers. 

## ***the research question***


 **Does prices differ bewteen supermarkets chains for the same product (with the same product code), in case of France, and if any supermarket is systematically cheaper?**

We choose France because it holds most of the data. Matching on barcode guarantees the products are identical, so a price gap reflects the retailer, not the product. The cell below checks that the data can answer this question.

```python
import os
import pandas as pd


# in order to download this data, set your directory, where you want to store it. For example, below is mine directory.
FOLDER = "/Users/lucjadudzinska/Desktop/dataanalysis in python /projekt"
LOCAL = os.path.join(FOLDER, "prices.parquet")
URL = "https://huggingface.co/datasets/openfoodfacts/open-prices/resolve/main/prices.parquet"

if os.path.exists(LOCAL):
    df = pd.read_parquet(LOCAL)
else:
    df = pd.read_parquet(URL)
    df.to_parquet(LOCAL)

df["price"] = pd.to_numeric(df["price"], errors="coerce")
df["date"] = pd.to_datetime(df["date"], errors="coerce")

print(f"Rows: {df.shape[0]:,}   Columns: {df.shape[1]}")
```

```python
# What one row is (for example)
print(df[["product_code", "product_name", "price", "currency",
          "location_osm_display_name", "date", "proof_type"]].head())


for col in df.columns:
    print( col, "-", df[col].dtype)


#data range
print("\nDate range:", df["date"].min().date(), "->", df["date"].max().date())
print("\nRows per year:")
print(df["date"].dt.year.dropna().astype(int).value_counts().sort_index())
```

```python
rows, cols = df.shape
total = rows * cols
missing = df.isna().sum().sum()

print(f"Rows:              {rows:,}")
print(f"Columns:           {cols:,}")
print(f"Total data points: {total:,}")
print(f"Missing values:    {missing:,} ({missing / total:.1%} of all data points)")

```
